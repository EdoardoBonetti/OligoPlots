````python



from Lab import SingleExperiment, Experiments, collect_files, plot_data
from os import listdir
%matplotlib inline
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
#CYBERPUNK = False
location = "DSC"

list_ids = collect_files(location, extension=".txt")
list_single_experiments = [SingleExperiment(id, location) for id in list_ids]
experiments = Experiments(list_single_experiments)
experiments.clean
experiments.flat_out

````

````python
experiments_H2 = experiments.filter(["heating", "2perc"])
plt.figure(figsize=(7, 4), dpi=1000)
title = "DSC thermogram oleogel RBX/RSO 2% heating"
plot_data(experiments_H2, title=title)
plt.legend(loc="upper right")
# plt.show()
plt.savefig("DSC_thermogram_heating_2perc.svg")#, dpi=1000)
#print(experiments_H2)
````

````python
experiments_C2 = experiments.filter(["cooling", "2perc"])
title = "DSC thermogram oleogel RBX/RSO 2% cooling"
plt.figure(figsize=(7, 4), dpi=1000)
plot_data(experiments_H2, title=title)
plt.legend(loc="upper right")
# plt.show()
plt.savefig("DSC_thermogram_cooling_2perc.svg")#, dpi=1000)
````

````python
experiments_H2 = experiments.filter(["heating", "2perc"])
experiments_H4 = experiments.filter(["heating", "4perc"])
experiments_H6 = experiments.filter(["heating", "6perc"])
experiments_H8 = experiments.filter(["heating", "8perc"])

plt.figure(figsize=(7, 4), dpi=1000)
plot_data(experiments_H8, starting_point=0.5 * 3, label="8%")
plot_data(experiments_H6, starting_point=0.5 * 2, label="6%")
plot_data(experiments_H4, starting_point=0.5 * 1, label="4%")
plot_data(experiments_H2, starting_point=0, label="2%")

plt.title("DSC thermogram RBX/RSO heating")
plt.legend(loc="upper right")
# plt.show()
plt.savefig("DSC_thermogram_heating_RBX.svg")
````

````python
experiments_C2 = experiments.filter(["cooling", "2perc"])
experiments_C4 = experiments.filter(["cooling", "4perc"])
experiments_C6 = experiments.filter(["cooling", "6perc"])
experiments_C8 = experiments.filter(["cooling", "8perc"])


plt.figure(figsize=(7, 4), dpi=1000)
plot_data(experiments_C8, starting_point=0.5 * 3, label="8%")
plot_data(experiments_C6, starting_point=0.5 * 2, label="6%")
plot_data(experiments_C4, starting_point=0.5 * 1, label="4%")
plot_data(experiments_C2, starting_point=0, label="2%")
plt.title("DSC thermogram oleogel RBX/RSO cooling")
plt.legend(loc="upper right")
# plt.show()
plt.savefig("DSC_thermogram_cooling_RBX.svg")#, dpi=1000)
````
