````python



from Lab import SingleExperiment, Experiments, collect_files, plot_data
from os import listdir
%matplotlib inline
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
CYBERPUNK = False
location = "LARD"

list_ids = collect_files(location, extension=".txt")
list_single_experiments = [SingleExperiment(id, location) for id in list_ids]
experiments = Experiments(list_single_experiments)
experiments.clean
experiments.flat_out

````

````python
experiments_H2 = experiments.filter(["heating"])
plt.figure(figsize=(7, 4), dpi=1000)
title = "DSC thermogram pork lard heating"
plot_data(experiments_H2, title=title, label="lard")
plt.legend(loc="upper right")
# plt.show()
plt.savefig("DSC_thermogram_heating_lard.svg")#, dpi=1000)
````

````python
experiments_C2 = experiments.filter(["cooling"])
plt.figure(figsize=(7, 4), dpi=1000)
title = "DSC thermogram pork lard cooling"
plot_data(experiments_C2, title=title, label="lard")
plt.legend(loc="upper right")
# plt.show()
plt.savefig("DSC_thermogram_cooling_lard.svg")#, dpi=1000)
````
